```python
top_product_window = Window.partitionBy("category").orderBy(col("bill_amount").desc())

top_products_df = final_joined_df.withColumn(
    "row_num",
    row_number().over(top_product_window)
).filter(
    col("row_num") == 1
)

top_products_df.select(
    "category",
    "product_name",
    "bill_amount"
).show()
```

```python
daily_orders_data = [
    (321,114,204,"2024-04-11",3,"Delivered"),
    (322,118,210,"2024-04-11",2,"Delivered"),
    (304,114,208,"2024-04-02",5,"Delivered"),
    (319,112,208,"2024-04-10",2,"Delivered"),
    (323,120,210,"2024-04-12",1,"Pending")
]

daily_orders_columns = [
    "order_id",
    "product_id",
    "supplier_id",
    "order_date",
    "quantity",
    "order_status"
]

daily_orders_df = spark.createDataFrame(
    daily_orders_data,
    daily_orders_columns
)

daily_orders_df.createOrReplaceTempView("daily_orders_updates")
```

```python
from delta.tables import DeltaTable

target_table = DeltaTable.forName(
    spark,
    "orders_merge_target"
)

target_table.alias("target").merge(
    daily_orders_df.alias("source"),
    "target.order_id = source.order_id"
).whenMatchedUpdate(set={
    "product_id": "source.product_id",
    "supplier_id": "source.supplier_id",
    "order_date": "source.order_date",
    "quantity": "source.quantity",
    "order_status": "source.order_status"
}).whenNotMatchedInsert(values={
    "order_id": "source.order_id",
    "product_id": "source.product_id",
    "supplier_id": "source.supplier_id",
    "order_date": "source.order_date",
    "quantity": "source.quantity",
    "order_status": "source.order_status"
}).execute()
```

It can update existing records and insert new records in a single operation.
MERGE helps maintain incremental data loads efficiently without rewriting the full table.
It is widely used in ETL pipelines and real-time data processing.

Parquet:
Supports only data storage, 
Does not support UPDATE or DELETE directly, 
No transaction history, 
No ACID transactions, 

Delta Lake:
Supports UPDATE, DELETE, MERGE
Supports Time Travel, 
Maintains transaction history, 
Provides ACID transactions, 
Better for ETL and streaming workloads.

```python
from delta.tables import DeltaTable

delta_table = DeltaTable.forPath(
    spark,
    "/tmp/products_delta"
)

delta_table.update(
    condition = "product_id = 111",
    set = {
        "price": "50000"
    }
)
```

Delta Lake supports updates better because it maintains a transaction log
called _delta_log. This provides ACID transactions, version control,
rollback capability, and efficient update operations without rewriting
the entire dataset.
